```python
corpus="""Hello welcome to the world of Natural Language Processing's Notebook.Tokenization is the process of breaking down text into smaller units called tokens, which can be words, phrases, or even sentences. 
"""
```

```python
print(corpus)
```

Tokenization. 


Sentence Tokenization

```python
from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

Paragraph-->Words.

Sentence-->>Words

```python
from nltk.tokenize import word_tokenize
```

```python
words=word_tokenize(corpus)
```

```python
words
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

Using Word_tokenizer ther the 's is one word. In wordpuct Tokenization ' and s is two other words

```python
wordpunct_tokenize(corpus)
```
