### `PorterStemmer`: 
It removes common prefixes and suffixes to produce the stem. The resulting stem may not always be a valid English word.

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

print(ps.stem("playing"))
print(ps.stem("studies"))
print(ps.stem("running"))
print(ps.stem("connected"))
```

`SnowballStemmer`:

It provides better accuracy and supports multiple languages.

```python
from nltk.stem import SnowballStemmer

ss = SnowballStemmer("english")

print(ss.stem("playing"))
print(ss.stem("studies"))
print(ss.stem("running"))
print(ss.stem("connected"))
```

```python
ps.stem("Fairly"),ps.stem("Sportingly")
```

```python
ss.stem("Fairly"),ss.stem("Sportingly")
```

`RegexpStemmer`: 

RegexpStemmer is a stemmer in NLTK that uses regular expressions (regex) to remove prefixes or suffixes from words.
Unlike Porter or Snowball Stemmer, it does not follow predefined linguistic rules. Instead, you define your own regex pattern for stemming.
It is useful when you want custom stemming based on specific requirements.

```python
from nltk.stem import RegexpStemmer

stemmer = RegexpStemmer('ing$|s$|ed$', min=4)

words = ["playing", "played", "plays", "running"]

for word in words:
    print(word, "->", stemmer.stem(word))
```

min=4 : This means:

If the word has 4 or more characters, apply stemming.
If it has fewer than 4 characters, leave it unchanged.
