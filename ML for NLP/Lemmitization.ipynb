`Lemmitization`

Lemmatization is the process of reducing a word to its base or dictionary form, called a lemma.
It uses vocabulary (dictionary) and grammatical (part-of-speech) information to find the correct root word.
Unlike stemming, the output of lemmatization is always a meaningful dictionary word.

`Basic Lemmatization`

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("cars"))
print(lemmatizer.lemmatize("books"))
print(lemmatizer.lemmatize("children"))
```

`Verb Lemmatization`

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("running", pos="v"))
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("better", pos="a"))   # adjective
print(lemmatizer.lemmatize("went", pos="v"))     # verb
print(lemmatizer.lemmatize("mice"))              # noun
```

```python
print(lemmatizer.lemmatize("Studies", pos="n"))  # noun
```

```python
print(lemmatizer.lemmatize("Studying", pos="v"))  # verb
```
